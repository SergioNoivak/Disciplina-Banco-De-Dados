```python
import os
import subprocess
from PyPDF2 import PdfMerger

# Caminho da pasta onde estão os arquivos
PASTA = os.getcwd()  # usa a pasta atual

def converter_pptx_para_pdf(pasta):
    """
    Converte todos os arquivos .pptx da pasta para PDF usando LibreOffice
    """
    for arquivo in sorted(os.listdir(pasta)):
        if arquivo.lower().endswith(".pptx"):
            caminho_completo = os.path.join(pasta, arquivo)
            print(f"Convertendo: {arquivo}")
            
            subprocess.run([
                "libreoffice",
                "--headless",
                "--convert-to",
                "pdf",
                caminho_completo,
                "--outdir",
                pasta
            ])

def juntar_pdfs(pasta, nome_saida="Aulas_BD_Completo.pdf"):
    """
    Junta todos os PDFs da pasta em um único arquivo
    """
    merger = PdfMerger()
    
    arquivos_pdf = sorted([
        f for f in os.listdir(pasta)
        if f.lower().endswith(".pdf")
    ])
    
    for pdf in arquivos_pdf:
        print(f"Adicionando ao PDF final: {pdf}")
        merger.append(os.path.join(pasta, pdf))
    
    caminho_saida = os.path.join(pasta, nome_saida)
    merger.write(caminho_saida)
    merger.close()
    
    print(f"\nPDF final criado com sucesso: {nome_saida}")

if __name__ == "__main__":
    print("=== Convertendo PPTX para PDF ===")
    converter_pptx_para_pdf(PASTA)
    
    print("\n=== Unindo PDFs ===")
    juntar_pdfs(PASTA)
    
    print("\nProcesso finalizado com sucesso!")
```
