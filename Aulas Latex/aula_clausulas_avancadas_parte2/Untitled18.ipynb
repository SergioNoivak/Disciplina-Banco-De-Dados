```python
%%sql

-- Create the log table if it doesn't exist
CREATE TABLE IF NOT EXISTS log_alunos_cursos (
    log_id INTEGER PRIMARY KEY AUTOINCREMENT,
    aluno_id INTEGER NOT NULL,
    aluno_nome TEXT NOT NULL,
    curso_antigo_id INTEGER,
    curso_novo_id INTEGER,
    data_modificacao DATETIME DEFAULT CURRENT_TIMESTAMP
);

-- Create the trigger to log course changes
CREATE TRIGGER IF NOT EXISTS log_de_mudanca_cursos
AFTER UPDATE OF curso ON Alunos
FOR EACH ROW
WHEN OLD.curso IS NOT NEW.curso
BEGIN
    INSERT INTO log_alunos_cursos (aluno_id, aluno_nome, curso_antigo_id, curso_novo_id)
    VALUES (OLD.id, OLD.nome, OLD.curso, NEW.curso);
END;
```

```python
%%sql
-- 1. Add operation_type column to the log table if it doesn't exist
-- This step ensures the log table can distinguish between different operation types.
ALTER TABLE log_alunos_cursos ADD COLUMN operation_type TEXT;

-- Drop existing update trigger to recreate it with the new column
DROP TRIGGER IF EXISTS log_de_mudanca_cursos;

-- 2. Create or modify existing AFTER UPDATE trigger
-- This trigger logs the old and new course IDs and marks the operation as 'UPDATE'.
CREATE TRIGGER IF NOT EXISTS log_de_mudanca_cursos
AFTER UPDATE OF curso ON Alunos
FOR EACH ROW
WHEN OLD.curso IS NOT NEW.curso
BEGIN
    INSERT INTO log_alunos_cursos (aluno_id, aluno_nome, curso_antigo_id, curso_novo_id, operation_type)
    VALUES (OLD.id, OLD.nome, OLD.curso, NEW.curso, 'UPDATE');
END;

-- 3. Create an AFTER INSERT trigger
-- This trigger logs details of a new student and marks the operation as 'INSERT'.
-- For INSERT, only NEW pseudo-row is available. OLD values are NULL.
CREATE TRIGGER IF NOT EXISTS log_de_insercao_alunos
AFTER INSERT ON Alunos
FOR EACH ROW
BEGIN
    INSERT INTO log_alunos_cursos (aluno_id, aluno_nome, curso_novo_id, operation_type)
    VALUES (NEW.id, NEW.nome, NEW.curso, 'INSERT');
END;

-- 4. Create an AFTER DELETE trigger
-- This trigger logs details of the deleted student and marks the operation as 'DELETE'.
-- For DELETE, only OLD pseudo-row is available. NEW values are NULL.
CREATE TRIGGER IF NOT EXISTS log_de_remocao_alunos
AFTER DELETE ON Alunos
FOR EACH ROW
BEGIN
    INSERT INTO log_alunos_cursos (aluno_id, aluno_nome, curso_antigo_id, operation_type)
    VALUES (OLD.id, OLD.nome, OLD.curso, 'DELETE');
END;


-- Demonstrating the triggers:

-- Check current log table state (optional)
SELECT * FROM log_alunos_cursos;

-- a) Example of INSERT operation (will fire log_de_insercao_alunos)
INSERT INTO Alunos (nome, idade, curso) VALUES ('Novo Aluno Teste', 20, 1);

-- b) Example of UPDATE operation (will fire log_de_mudanca_cursos)
-- Note: An existing log entry from the previous step might already be there for Aluno ID 1.
-- This update will create a new entry in the log.
UPDATE Alunos SET curso = 3 WHERE id = 1; -- Changing João Silva's course again

-- c) Example of DELETE operation (will fire log_de_remocao_alunos)
DELETE FROM Alunos WHERE nome = 'Novo Aluno Teste';

-- Verify the log table after all operations
SELECT * FROM log_alunos_cursos;
```
